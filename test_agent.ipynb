```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

coin = "Bitcoin (BTC)"
price_change = "+6% in the last 2 hours"
news = "Rumors of a major exchange facing technical issues and delaying withdrawals."

system_prompt = "You are a cautious crypto risk manager. Answer in 2 short sentences."

user_prompt = f"""
Coin: {coin}
Price movement: {price_change}
Latest news: {news}

Should my trading bot enter a BUY trade right now? Explain why.
"""

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt}
    ]
)

print(response.choices.message.content)
```
